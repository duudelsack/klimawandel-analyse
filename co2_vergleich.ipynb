```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Verzeichnis mit den CSV-Dateien
data_dir = './data/'

# Liste der zu vergleichenden Sektoren
sectors_to_use = ['power', 'transportation', 'waste', 'buildings']

# Dictionary zum Speichern der DataFrames
data = {}

# Alle Dateien im Datenordner durchsuchen
for filename in os.listdir(data_dir):
    for sector in sectors_to_use:
        if sector in filename.lower() and filename.endswith('.csv'):
            sector_name = sector.capitalize()
            filepath = os.path.join(data_dir, filename)
            df = pd.read_csv(filepath)

            # Anpassung je nach Struktur der CSVs
            if 'Year' in df.columns and 'Emissions' in df.columns:
                df = df[['Year', 'Emissions']].copy()
            elif 'year' in df.columns and 'emissions' in df.columns:
                df = df[['year', 'emissions']].copy()
                df.columns = ['Year', 'Emissions']
            else:
                continue  # überspringen, wenn Spalten nicht passen

            df.rename(columns={'Emissions': sector_name}, inplace=True)
            data[sector_name] = df
            break  # Wenn die Datei einem Sektor zugeordnet wurde, nächste Datei prüfen

# Daten zusammenführen
merged_df = None
for sector, df in data.items():
    if merged_df is None:
        merged_df = df
    else:
        merged_df = pd.merge(merged_df, df, on='Year', how='outer')

# Nach Jahr sortieren
merged_df.sort_values('Year', inplace=True)

# Plot
plt.figure(figsize=(12, 6))
for sector in data.keys():
    plt.plot(merged_df['Year'], merged_df[sector], label=sector)

plt.title('CO₂-Emissionen nach Sektor über die Jahre')
plt.xlabel('Jahr')
plt.ylabel('Emissionen (in Mt CO₂e)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
